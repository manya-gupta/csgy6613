# Model usage

```python
from ultralytics import YOLO

model = YOLO('/content/drive/MyDrive/AI2026spring/hw2/best.pt')
```

```python
import pandas as pd
class_names = model.names
qdf = pd.DataFrame(columns = list(class_names.values()))
```

```python
from pathlib import Path

# Define the directory path
directory_path = Path("/content/drive/MyDrive/AI2026spring/hw2/frames")
frames_folder = "/content/drive/MyDrive/AI2026spring/hw2/frames/"

# Get a list of file names using list comprehension and is_file()
file_names = [p.name for p in directory_path.iterdir() if p.is_file()]

qdf.insert(23, 'path', file_names)
```

```python
out_df = pd.DataFrame(columns = ("VID", "Frame Index", "Class Label", "Bounding Box", "Confidence Score"))
video_id = 'YcvECxtXoxQ'
```

```python
for i, path in enumerate(qdf['path']):
  image = Path(frames_folder + path)
  results = model(image, imgsz=640)

  for result in results:
    boxes = result.boxes  # Boxes object for bounding box outputs

  cls_list = result.boxes.cls.tolist()
  cls_idx_list = [int(item) for item in cls_list]
  for cls in cls_idx_list:
    qdf.iat[i,cls] = 1

  cls_label_list = [class_names[item] for item in cls_idx_list]

  out_df.loc[i, 'VID'] = video_id
  out_df.loc[i, 'Frame Index'] = i

  out_df.loc[i, 'Class Label'] = cls_label_list
  out_df.loc[i, 'Bounding Box'] = result.boxes.xyxyn.tolist()
  out_df.loc[i, 'Confidence Score'] = result.boxes.conf.tolist()

qdf.fillna(0, inplace=True)
```

# Semantic Search

The process image function should work with any input image_path and any model. I used a pre-trained YOLO model with some fine-tuning.

```python
from datetime import datetime
import torch
import time

'''
inputs: image_path, model
outputs: time_log, number_of_supporting_detections
'''
def process_image(image_path, model):
  # Run batched inference on a list of images
  results = model(image_path,  imgsz=640, conf=0.25)  # return a list of Results objects

  # Process results list
  for result in results:
      boxes = result.boxes  # Boxes object for bounding box outputs
      result.show()  # display to screen

  cls_tensor = result.boxes.cls
  cls_list = cls_tensor.tolist()
  cls_idx_list = [int(item) for item in cls_list]

  # list of named elements detected in input image
  cls_label_list = [class_names[item] for item in cls_idx_list]

  time_log = pd.DataFrame(columns = ["class_label", "start_time", "end_time"])

  for element in cls_label_list:
    frame_list = pd.Series(qdf[element]) # element is stored as frame_list.name
    started = False
    for frame, flag in enumerate(frame_list):
      if flag == 1 and not started:
        start_time = time.strftime('%M:%S', time.gmtime(frame * 3))
        started = True
      elif flag == 0 and started:
        end_time = time.strftime('%M:%S', time.gmtime(frame * 3))
        time_log.loc[len(time_log)] = [element, start_time, end_time]
        started = False

  return time_log, len(time_log)

timelog, number_of_supporting_detections = process_image(frames_folder + "frame_0001.jpg", model)
```

```python
print(timelog)
print(f"Number of supporting detections: {number_of_supporting_detections}")
```
